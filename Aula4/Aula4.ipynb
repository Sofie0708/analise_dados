```python
import pandas as pd
import openpyxl
```

```python
path = r"H:\python\python-tecTI\dados\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
tabela_bruta = pd.read_excel(path, header=None)
```

```python
tabela_bruta
```

```python
tabela_censo = tabela_bruta.iloc[3:5572, [1,4,5,7]].copy()
```

```python
tabela_censo
```

```python
tabela_censo.columns = [
    "UF",
    "Municipio",
    "pop10",
    "pop22"
]
```

```python
tabela_censo["pop10"] = pd.to_numeric(tabela_censo["pop10"], errors="coerce")
tabela_censo["pop22"] = pd.to_numeric(tabela_censo["pop22"], errors="coerce")
```

```python
tabela_censo["dif"] = tabela_censo["pop22"] - tabela_censo["pop10"]
```

```python
pop_estado = (
    tabela_censo.groupby(["UF"], as_index=False)
      .agg(
          pop_2010=("pop10", "sum"),
          pop_2022=("pop22", "sum"),
          dif_estado = ("dif", "sum")
          
      )
)
```

```python
pop_estado = pop_estado.sort_values(by="dif_estado", ascending=False, ignore_index=True)
```

```python
pop_estado
```

```python
pop_municipio = tabela_censo.sort_values(by="dif", ascending=False, ignore_index=True)
```

```python
pop_municipio.to_csv(r"H:\python\python-tecTI\dados\Dados_municipio.csv")
```
